**Q1** Describe the data you see here. How many variables and observations. What are the data types? What are the units?

It seems to be tidy. There are 12 observations across 5 variables, but the variables aren't really distinct. Because they are proportions, this is really two variables (religion and income). And the observations are the sample size.

**Q2** Following along with the lecture, say in words what each section of the code below does:

**(Click here to answer each one)**

* Section 1 - Renaming the columns "Religious tradition" and "Sample size" to "religion" and "n", respectively.
* Section 2 - The melt function is changing the way our data frame is arranged such that income and proportion are added as columns, and the values are found from the cooresponding spot in the origional data frame. So instead of sorting my income as a distribution, the income type and the proportion of that religion that have that income are now their own variable columns.
* Section 3 - This is creating a new column 'frequency' that is the proportion times the sample size in order to get a value of how many people are in that income braket per religion. It also is a rounded integer (since there can't be a fraction of a person).
* Section 4 - This uses the lambda function to overright our data frame and replaces the dollar sign with just a space (it looks like in order to turn this string into an integer so we can find information about it).
* Section 5 - Replacing the index to the one that it is currently sitting in. It adds it as a secondary index so we have the original, but also have a working index for our new sorting.
* Section 6 - Changing the name from "Unaffiliated (religions "nones")" to "Unaffiliated"
* Section 7 - Changing the name "Historically Black Protestant" to "Hist. Black Protestant"
* Section 8 - Changing the name from "Evangelical Protestant" to "Ev. Protestant"

````python
# Section 1
DF.rename(columns={ 'Religious tradition':'religion' ,'Sample Size' : 'n' },inplace=True)
# Section 2
DF_new = pd.melt(DF, id_vars=['religion','n'], var_name='income', value_name='proportion').sort_values('religion')
# Section 3
DF_new['frequency']=np.round(DF_new['proportion']*DF_new['n'])
# Section 4
DF_new['income']=DF_new['income'].apply(lambda x: str(x).replace('$',''))
# Section 5
DF_new.reset_index(inplace=True)

# Section 6
name_to_change = 'Unaffiliated (religious "nones")'
new_name = 'Unaffiliated'
DF_new['religion']=DF_new['religion'].replace(name_to_change,new_name)

# Section 7
name_to_change =  'Historically Black Protestant'
new_name = 'Hist. Black Protestant'
DF_new['religion']=DF_new['religion'].replace(name_to_change,new_name)

# Section 8
name_to_change = 'Evangelical Protestant'
new_name = 'Ev. Protestant'
DF_new['religion']=DF_new['religion'].replace(name_to_change,new_name)

DF_new
````

````python

fig = px.bar(DF_new,
             y='religion',
             x='proportion',
             color='income',
             pattern_shape='income',
             color_discrete_sequence=px.colors.qualitative.Vivid,
             category_orders={'income' : ['Less than 30,000', '30,000-49,999', '50,000-99,999', '100,000 or more']})

fig.update_layout(template="plotly_dark",
                  title='Income Distribution by religious group <br><sup>Data Source: Pew Research Center, Religious Lansdcape Study</sup>',
                  title_x=0.5,
                  yaxis={'categoryorder': 'category descending'},
                  xaxis_title="Proportion",
                  yaxis_title="",
                  legend_title='Income',
                  legend={'orientation':"h",'yanchor':"top",'y':-0.2, 'xanchor':"right",'x':1.05},
                  font={'family':"Overpass, monospace",'size':16,'color':"white"},
                  autosize=False,
                  width=1000,
                  height=800)

fig.show()

````
